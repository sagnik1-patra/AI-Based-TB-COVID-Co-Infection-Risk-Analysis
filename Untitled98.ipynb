```python
import os
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_predict
)

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.pipeline import Pipeline

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based TB-COVID Co-Infection Risk Analysis"
)

DATA_PATH = os.path.join(
    BASE_DIR,
    "2.13_TB_COVID.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 120

print("=" * 75)
print("AI-BASED TB-COVID CO-INFECTION RISK ANALYSIS")
print("=" * 75)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(
    DATA_PATH,
    encoding="utf-8-sig",
    index_col=False
)

df.columns = df.columns.str.strip()

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

print("\nFirst five records:")
print(df.head())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

# Expected 13-column TB-COVID dataset.
# Verify this mapping against your CSV headers
# before running if your file format differs.

if df.shape[1] != 13:
    raise ValueError(
        "Expected 13 columns but found "
        f"{df.shape[1]}."
    )

df.columns = [
    "State_UT",
    "Public_Screened_Count",
    "Public_Screening_Percentage",
    "Private_Screened_Count",
    "Private_Screening_Percentage",
    "Total_Screened_Count",
    "Total_Screening_Percentage",
    "Public_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detected_Count",
    "Private_Detection_Percentage",
    "Total_Detected_Count",
    "Total_Detection_Percentage"
]

# ============================================================
# 4. DATA CLEANING
# ============================================================

df["State_UT"] = (
    df["State_UT"]
    .astype(str)
    .str.strip()
)

numeric_columns = [
    col for col in df.columns
    if col != "State_UT"
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

print("\nMissing values:")
print(df.isnull().sum())

# Remove missing target values
df = df.dropna(
    subset=["Total_Screening_Percentage"]
).copy()

# Remove aggregate totals
aggregate_names = {
    "india",
    "all india",
    "total",
    "grand total"
}

df = df[
    ~df["State_UT"]
    .str.lower()
    .isin(aggregate_names)
].copy()

df.reset_index(drop=True, inplace=True)

print("\nCleaned dataset shape:", df.shape)

# ============================================================
# 5. CREATE CLASSIFICATION TARGET
# ============================================================

# These are illustrative screening coverage bands.
# They are not clinical risk categories.

def classify_screening(value):

    if value < 50:
        return "Low"

    elif value < 80:
        return "Moderate"

    return "High"


df["Screening_Performance_Category"] = (
    df["Total_Screening_Percentage"]
    .apply(classify_screening)
)

print("\nClass distribution:")
print(
    df["Screening_Performance_Category"]
    .value_counts()
)

# ============================================================
# 6. SELECT FEATURES
# ============================================================

# Exclude the target-defining percentage and
# related screening percentages to reduce
# direct target leakage.
#
# This is exploratory state-level classification,
# not individual patient risk prediction.

feature_columns = [
    "Public_Screened_Count",
    "Private_Screened_Count",
    "Total_Screened_Count",
    "Public_Detected_Count",
    "Private_Detected_Count",
    "Total_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detection_Percentage",
    "Total_Detection_Percentage"
]

X = df[feature_columns].copy()

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df["Screening_Performance_Category"]
)

class_names = label_encoder.classes_.tolist()
num_classes = len(class_names)

print("\nClasses:", class_names)
print("\nSelected features:", feature_columns)

if num_classes < 2:
    raise ValueError(
        "At least two screening categories "
        "are required for classification."
    )

# ============================================================
# 7. TRAIN-TEST SPLIT
# ============================================================

class_counts = np.bincount(y)

test_count = max(
    num_classes,
    int(np.ceil(len(y) * 0.25))
)

if test_count >= len(y):
    raise ValueError(
        "Not enough observations for evaluation."
    )

stratify_target = None

if (
    class_counts.min() >= 2
    and len(y) - test_count >= num_classes
):
    stratify_target = y

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=test_count,
        random_state=RANDOM_STATE,
        stratify=stratify_target
    )
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 8. DEFINE MACHINE LEARNING MODELS
# ============================================================

models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=5,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=4,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "SVM": SVC(
        kernel="rbf",
        probability=True,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=3
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        random_state=RANDOM_STATE
    )
}

# ============================================================
# 9. TRAIN AND EVALUATE MODELS
# ============================================================

results = []
trained_models = {}
test_predictions = {}

for name, estimator in models.items():

    print("\n" + "-" * 60)
    print("Training:", name)

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            estimator
        )
    ])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    trained_models[name] = pipeline
    test_predictions[name] = predictions

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

# ============================================================
# 10. MODEL COMPARISON DATAFRAME
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_path = os.path.join(
    BASE_DIR,
    "model_comparison.csv"
)

comparison_df.to_csv(
    comparison_path,
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

# ============================================================
# 11. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    hue="Model",
    palette="viridis",
    legend=False
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

plt.title(
    "Accuracy Comparison of Machine Learning Models",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")
plt.ylim(0, 1.12)
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

accuracy_path = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

plt.savefig(
    accuracy_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nSaved:", accuracy_path)

# ============================================================
# 12. CORRELATION HEATMAP
# ============================================================

heatmap_columns = (
    feature_columns +
    ["Total_Screening_Percentage"]
)

correlation_matrix = (
    df[heatmap_columns]
    .corr(numeric_only=True)
)

plt.figure(figsize=(15, 11))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": 0.8}
)

plt.title(
    "TB-COVID Screening and Detection Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()

heatmap_path = os.path.join(
    BASE_DIR,
    "heatmap.png"
)

plt.savefig(
    heatmap_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", heatmap_path)

# ============================================================
# 13. MODEL COMPARISON GRAPH
# ============================================================

comparison_long = comparison_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=comparison_long,
    x="Model",
    y="Score",
    hue="Metric",
    palette="Set2"
)

plt.title(
    "Machine Learning Model Performance Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=30, ha="right")
plt.legend(title="Evaluation Metric")
plt.tight_layout()

comparison_graph_path = os.path.join(
    BASE_DIR,
    "comparison_graph.png"
)

plt.savefig(
    comparison_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", comparison_graph_path)

# ============================================================
# 14. SELECT MODEL FOR RESULT EXPORT
# ============================================================

# Use Random Forest as the primary reporting model.
# This avoids choosing the model using test-set
# performance and then reporting that same test
# performance as an unbiased selected-model result.

best_model_name = "Random Forest"

best_model = trained_models[best_model_name]

best_predictions = test_predictions[best_model_name]

print("\nPrimary reporting model:", best_model_name)

# ============================================================
# 15. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        best_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 16. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    best_predictions,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "Random Forest Confusion Matrix",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")
plt.tight_layout()

confusion_path = os.path.join(
    BASE_DIR,
    "confusion_matrix.png"
)

plt.savefig(
    confusion_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", confusion_path)

# ============================================================
# 17. CREATE RESULT CSV
# ============================================================

# These are held-out test-set results.
# They are distinct from predictions on all rows.

result_df = df.loc[
    X_test.index,
    [
        "State_UT",
        "Total_Screening_Percentage",
        "Total_Detection_Percentage"
    ]
].copy()

result_df["Actual_Category"] = (
    label_encoder.inverse_transform(y_test)
)

result_df["Predicted_Category"] = (
    label_encoder.inverse_transform(
        best_predictions
    )
)

result_df["Actual_Encoded"] = y_test
result_df["Predicted_Encoded"] = best_predictions

result_df["Correct_Prediction"] = (
    result_df["Actual_Category"]
    == result_df["Predicted_Category"]
)

result_df = result_df.sort_values(
    by="State_UT"
).reset_index(drop=True)

result_path = os.path.join(
    BASE_DIR,
    "result.csv"
)

result_df.to_csv(
    result_path,
    index=False
)

print("\nRESULT CSV:")
print(result_df.to_string(index=False))

print("\nSaved:", result_path)

# ============================================================
# 18. RESULT GRAPH
# ============================================================

plt.figure(figsize=(14, 7))

positions = np.arange(len(result_df))

plt.plot(
    positions,
    result_df["Actual_Encoded"],
    marker="o",
    linewidth=2,
    label="Actual Category"
)

plt.plot(
    positions,
    result_df["Predicted_Encoded"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Predicted Category"
)

plt.title(
    "Actual vs Predicted Screening Performance",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Screening Performance Category")

plt.xticks(
    positions,
    result_df["State_UT"],
    rotation=75,
    ha="right"
)

plt.yticks(
    range(num_classes),
    class_names
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

result_graph_path = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

plt.savefig(
    result_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", result_graph_path)

# ============================================================
# 19. REFIT PRIMARY MODEL FOR ALL-STATE PREDICTIONS
# ============================================================

# The final model is fitted on all available states
# after the held-out evaluation has been completed.
#
# These full-dataset predictions are descriptive
# fitted predictions, not independent test results.

final_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=100,
            max_depth=5,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=RANDOM_STATE
        )
    )
])

final_model.fit(X, y)

all_predictions = final_model.predict(X)

all_probabilities = final_model.predict_proba(X)

# Probability column order comes from
# the classifier's classes_ attribute.

classifier_classes = (
    final_model.named_steps["classifier"].classes_
)

probability_columns = {}

for column_index, class_id in enumerate(
    classifier_classes
):

    category_name = (
        label_encoder.inverse_transform(
            [class_id]
        )[0]
    )

    probability_columns[
        f"Probability_{category_name}"
    ] = all_probabilities[:, column_index]

# ============================================================
# 20. CREATE PREDICTION CSV
# ============================================================

prediction_df = df[
    [
        "State_UT",
        "Total_Screening_Percentage",
        "Total_Detection_Percentage"
    ]
].copy()

prediction_df["Predicted_Category"] = (
    label_encoder.inverse_transform(
        all_predictions
    )
)

prediction_df["Predicted_Encoded"] = (
    all_predictions
)

prediction_df["Prediction_Confidence"] = (
    np.max(all_probabilities, axis=1)
)

for column_name, values in probability_columns.items():

    prediction_df[column_name] = values

prediction_df = prediction_df.sort_values(
    by="State_UT"
).reset_index(drop=True)

prediction_path = os.path.join(
    BASE_DIR,
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("\nPREDICTION CSV:")
print(prediction_df.to_string(index=False))

print("\nSaved:", prediction_path)

# ============================================================
# 21. PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(15, 8))

category_colors = {
    "Low": "tomato",
    "Moderate": "orange",
    "High": "seagreen"
}

bar_colors = [
    category_colors.get(category, "steelblue")
    for category in prediction_df["Predicted_Category"]
]

plt.bar(
    prediction_df["State_UT"],
    prediction_df["Predicted_Encoded"],
    color=bar_colors,
    alpha=0.85
)

plt.title(
    "Predicted TB-COVID Screening Performance by State",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Predicted Screening Category")

plt.yticks(
    range(num_classes),
    class_names
)

plt.xticks(
    rotation=85,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()

prediction_graph_path = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)

plt.savefig(
    prediction_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", prediction_graph_path)

# ============================================================
# 22. CLASS DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(9, 6))

class_distribution = (
    df["Screening_Performance_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

ax = sns.barplot(
    x=class_distribution.index,
    y=class_distribution.values,
    hue=class_distribution.index,
    palette="Set2",
    legend=False
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%d",
        padding=3
    )

plt.title(
    "State-Level Screening Performance Distribution",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Screening Performance Category")
plt.ylabel("Number of States / UTs")
plt.tight_layout()

distribution_path = os.path.join(
    BASE_DIR,
    "class_distribution_graph.png"
)

plt.savefig(
    distribution_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", distribution_path)

# ============================================================
# 23. FEATURE IMPORTANCE GRAPH
# ============================================================

rf_classifier = (
    final_model.named_steps["classifier"]
)

feature_importance = pd.DataFrame({
    "Feature": feature_columns,
    "Importance": rf_classifier.feature_importances_
})

feature_importance = (
    feature_importance.sort_values(
        "Importance",
        ascending=False
    )
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature",
    hue="Feature",
    palette="viridis",
    legend=False
)

plt.title(
    "Random Forest Feature Importance",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.tight_layout()

importance_path = os.path.join(
    BASE_DIR,
    "feature_importance_graph.png"
)

plt.savefig(
    importance_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved:", importance_path)

# ============================================================
# 24. SAVE FEATURE IMPORTANCE CSV
# ============================================================

feature_importance_path = os.path.join(
    BASE_DIR,
    "feature_importance.csv"
)

feature_importance.to_csv(
    feature_importance_path,
    index=False
)

print("\nSaved:", feature_importance_path)

# ============================================================
# 25. SAVE FINAL TRAINED MODEL
# ============================================================

model_package = {
    "model": final_model,
    "label_encoder": label_encoder,
    "feature_columns": feature_columns,
    "class_names": class_names,
    "target_column":
        "Screening_Performance_Category"
}

model_path = os.path.join(
    BASE_DIR,
    "tb_covid_model.pkl"
)

joblib.dump(
    model_package,
    model_path
)

print("\nFinal trained model saved:")
print(model_path)

# ============================================================
# 26. VERIFY OUTPUT FILES
# ============================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "model_comparison.csv",
    "class_distribution_graph.png",
    "feature_importance_graph.png",
    "feature_importance.csv",
    "tb_covid_model.pkl"
]

print("\n" + "=" * 75)
print("VERIFYING GENERATED FILES")
print("=" * 75)

for filename in output_files:

    file_path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(file_path):

        size_kb = (
            os.path.getsize(file_path) / 1024
        )

        print(
            f"[SUCCESS] {filename} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[MISSING] {filename}"
        )

# ============================================================
# 27. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("PROJECT EXECUTION COMPLETED")
print("=" * 75)

print("\nModel Comparison:")
print(comparison_df.to_string(index=False))

print("\nPrimary reporting model:")
print(best_model_name)

print("\nHeld-out Test Accuracy:")
print(
    round(
        accuracy_score(y_test, best_predictions),
        4
    )
)

print("\nAll graphs and CSV files saved in:")
print(BASE_DIR)

print("\nIMPORTANT:")
print(
    "Predictions are exploratory state-level "
    "screening classifications, not clinical "
    "co-infection risk predictions."
)

print("\nDONE!")
```
