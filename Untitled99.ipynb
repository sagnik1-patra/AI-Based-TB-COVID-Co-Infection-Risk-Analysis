```python
import os
import json
import copy
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based TB-COVID Co-Infection Risk Analysis"
)

DATA_PATH = os.path.join(
    BASE_DIR,
    "2.13_TB_COVID.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

SEED = 42
rng = np.random.default_rng(SEED)

np.random.seed(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 120

def output_path(filename):
    return os.path.join(
        BASE_DIR,
        "ais_" + filename
    )

def save_show(filename):
    path = output_path(filename)

    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

    print("Saved:", path)

print("=" * 75)
print("AIS-BASED TB-COVID CO-INFECTION ANALYSIS")
print("=" * 75)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(DATA_PATH)

df = pd.read_csv(
    DATA_PATH,
    encoding="utf-8-sig",
    index_col=False
)

df.columns = df.columns.str.strip()

print("\nOriginal dataset shape:", df.shape)
print(df.head())

# ============================================================
# 3. STANDARDIZE COLUMNS
# ============================================================

# Expected dataset: 13 columns.
# This positional mapping must match the actual CSV.

if df.shape[1] != 13:
    raise ValueError(
        "Expected 13 columns, found "
        f"{df.shape[1]}"
    )

df.columns = [
    "State_UT",
    "Public_Screened_Count",
    "Public_Screening_Percentage",
    "Private_Screened_Count",
    "Private_Screening_Percentage",
    "Total_Screened_Count",
    "Total_Screening_Percentage",
    "Public_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detected_Count",
    "Private_Detection_Percentage",
    "Total_Detected_Count",
    "Total_Detection_Percentage"
]

# ============================================================
# 4. DATA CLEANING
# ============================================================

df["State_UT"] = (
    df["State_UT"]
    .astype(str)
    .str.strip()
)

numeric_columns = [
    col for col in df.columns
    if col != "State_UT"
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

df = df.dropna(
    subset=["Total_Screening_Percentage"]
).copy()

aggregate_names = {
    "india",
    "all india",
    "total",
    "grand total"
}

df = df[
    ~df["State_UT"]
    .str.lower()
    .isin(aggregate_names)
].copy()

df.reset_index(drop=True, inplace=True)

print("\nCleaned dataset:", df.shape)

# ============================================================
# 5. CREATE TARGET CATEGORIES
# ============================================================

# Screening performance categories are illustrative.
# They are not clinical co-infection risk categories.

def screening_category(value):

    if value < 50:
        return "Low"

    elif value < 80:
        return "Moderate"

    return "High"

df["Screening_Performance_Category"] = (
    df["Total_Screening_Percentage"]
    .apply(screening_category)
)

encoder = LabelEncoder()

y = encoder.fit_transform(
    df["Screening_Performance_Category"]
)

class_names = encoder.classes_.tolist()
num_classes = len(class_names)

print("\nClass distribution:")
print(
    df["Screening_Performance_Category"]
    .value_counts()
)

if num_classes < 2:
    raise ValueError(
        "At least two classes are required."
    )

# ============================================================
# 6. FEATURE SELECTION INPUT
# ============================================================

# Do not include the target-defining screening
# percentage or the related screening percentages.
#
# This is exploratory state-level analysis.

feature_columns = [
    "Public_Screened_Count",
    "Private_Screened_Count",
    "Total_Screened_Count",
    "Public_Detected_Count",
    "Private_Detected_Count",
    "Total_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detection_Percentage",
    "Total_Detection_Percentage"
]

X = df[feature_columns].copy()

n_features = len(feature_columns)

# ============================================================
# 7. TRAIN-TEST SPLIT
# ============================================================

class_counts = np.bincount(y)

test_count = max(
    num_classes,
    int(np.ceil(len(y) * 0.25))
)

if test_count >= len(y):
    raise ValueError(
        "Not enough records for evaluation."
    )

stratify_target = None

if (
    class_counts.min() >= 2
    and len(y) - test_count >= num_classes
):
    stratify_target = y

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=test_count,
        random_state=SEED,
        stratify=stratify_target
    )
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# ============================================================
# 8. AIS CONFIGURATION
# ============================================================

POPULATION_SIZE = 12
GENERATIONS = 12
TOP_ANTIBODIES = 4
CLONES_PER_ANTIBODY = 4
RANDOM_NEW_ANTIBODIES = 3

# Antibody representation:
# First n_features values: binary feature mask
# Next value: n_estimators
# Next value: max_depth
# Next value: min_samples_leaf

def create_antibody():

    mask = rng.integers(
        0, 2,
        size=n_features
    )

    if mask.sum() == 0:
        mask[rng.integers(n_features)] = 1

    return {
        "mask": mask,
        "n_estimators": int(
            rng.integers(40, 151)
        ),
        "max_depth": int(
            rng.integers(2, 11)
        ),
        "min_samples_leaf": int(
            rng.integers(1, 5)
        )
    }

# ============================================================
# 9. FITNESS FUNCTION
# ============================================================

# Fitness uses only the training set.
# The test set is not used during optimization.
#
# With a very small dataset, CV results can be
# unstable. Feature count penalty encourages
# simpler solutions.

train_counts = np.bincount(
    y_train,
    minlength=num_classes
)

min_train_class_count = int(
    train_counts[train_counts > 0].min()
)

CV_FOLDS = min(
    3,
    min_train_class_count
)

if CV_FOLDS < 2:
    raise ValueError(
        "Not enough training examples per class "
        "for stratified AIS optimization. "
        "More observations are required."
    )

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def antibody_key(antibody):

    return (
        tuple(antibody["mask"].tolist()),
        antibody["n_estimators"],
        antibody["max_depth"],
        antibody["min_samples_leaf"]
    )

def fitness(antibody):

    key = antibody_key(antibody)

    if key in fitness_cache:
        return fitness_cache[key]

    selected_indices = np.where(
        antibody["mask"] == 1
    )[0]

    if len(selected_indices) == 0:
        return -1.0

    selected_features = [
        feature_columns[i]
        for i in selected_indices
    ]

    X_subset = X_train[
        selected_features
    ]

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=antibody["n_estimators"],
                max_depth=antibody["max_depth"],
                min_samples_leaf=antibody["min_samples_leaf"],
                class_weight="balanced",
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

    try:

        scores = cross_val_score(
            pipeline,
            X_subset,
            y_train,
            cv=cv,
            scoring="f1_macro",
            n_jobs=1,
            error_score="raise"
        )

        mean_score = float(
            np.mean(scores)
        )

    except ValueError:

        mean_score = 0.0

    feature_penalty = (
        0.02 *
        len(selected_features) /
        n_features
    )

    final_fitness = (
        mean_score - feature_penalty
    )

    fitness_cache[key] = final_fitness

    return final_fitness

# ============================================================
# 10. CLONAL MUTATION
# ============================================================

def mutate_antibody(
    antibody,
    mutation_rate=0.2
):

    clone = copy.deepcopy(antibody)

    for i in range(n_features):

        if rng.random() < mutation_rate:

            clone["mask"][i] = (
                1 - clone["mask"][i]
            )

    if clone["mask"].sum() == 0:

        index = rng.integers(n_features)
        clone["mask"][index] = 1

    if rng.random() < mutation_rate:

        clone["n_estimators"] = int(
            np.clip(
                clone["n_estimators"]
                + rng.integers(-25, 26),
                40,
                150
            )
        )

    if rng.random() < mutation_rate:

        clone["max_depth"] = int(
            np.clip(
                clone["max_depth"]
                + rng.integers(-2, 3),
                2,
                10
            )
        )

    if rng.random() < mutation_rate:

        clone["min_samples_leaf"] = int(
            np.clip(
                clone["min_samples_leaf"]
                + rng.integers(-1, 2),
                1,
                4
            )
        )

    return clone

# ============================================================
# 11. AIS OPTIMIZATION
# ============================================================

print("\nStarting AIS Optimization...")

population = [
    create_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = -np.inf

fitness_history = []

for generation in range(GENERATIONS):

    evaluated = [
        (fitness(a), a)
        for a in population
    ]

    evaluated.sort(
        key=lambda item: item[0],
        reverse=True
    )

    generation_best_score = (
        evaluated[0][0]
    )

    if generation_best_score > best_fitness:

        best_fitness = generation_best_score

        best_antibody = copy.deepcopy(
            evaluated[0][1]
        )

    fitness_history.append(best_fitness)

    print(
        f"Generation {generation + 1}/{GENERATIONS}"
        f" | Best Fitness: {best_fitness:.4f}"
    )

    elite = [
        copy.deepcopy(item[1])
        for item in evaluated[:TOP_ANTIBODIES]
    ]

    clones = []

    for rank, antibody in enumerate(elite):

        # Better antibodies receive lower
        # mutation rates.

        mutation_rate = (
            0.08 + rank * 0.07
        )

        for _ in range(CLONES_PER_ANTIBODY):

            clone = mutate_antibody(
                antibody,
                mutation_rate
            )

            clones.append(clone)

    candidates = elite + clones

    candidate_scores = [
        (fitness(a), a)
        for a in candidates
    ]

    candidate_scores.sort(
        key=lambda item: item[0],
        reverse=True
    )

    population = [
        copy.deepcopy(item[1])
        for item in candidate_scores[
            :POPULATION_SIZE - RANDOM_NEW_ANTIBODIES
        ]
    ]

    population += [
        create_antibody()
        for _ in range(RANDOM_NEW_ANTIBODIES)
    ]

print("\nAIS Optimization Completed!")

print("Best fitness:", best_fitness)
print("Best antibody:", best_antibody)

# ============================================================
# 12. SELECT OPTIMIZED FEATURES
# ============================================================

selected_indices = np.where(
    best_antibody["mask"] == 1
)[0]

selected_features = [
    feature_columns[i]
    for i in selected_indices
]

print("\nAIS Selected Features:")

for feature in selected_features:
    print("-", feature)

selected_features_df = pd.DataFrame({
    "Feature": feature_columns,
    "Selected": [
        int(value)
        for value in best_antibody["mask"]
    ]
})

selected_features_df.to_csv(
    output_path("selected_features.csv"),
    index=False
)

# ============================================================
# 13. OPTIMIZATION CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    range(1, GENERATIONS + 1),
    fitness_history,
    marker="o",
    linewidth=2
)

plt.title(
    "AIS Clonal Selection Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Best Fitness")
plt.grid(alpha=0.3)

save_show("optimization_graph.png")

# ============================================================
# 14. DEFINE MODELS
# ============================================================

models = {
    "AIS Random Forest": RandomForestClassifier(
        n_estimators=best_antibody["n_estimators"],
        max_depth=best_antibody["max_depth"],
        min_samples_leaf=best_antibody["min_samples_leaf"],
        class_weight="balanced",
        random_state=SEED
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=4,
        class_weight="balanced",
        random_state=SEED
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=SEED
    ),

    "SVM": SVC(
        probability=True,
        class_weight="balanced",
        random_state=SEED
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=3
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        random_state=SEED
    )
}

# ============================================================
# 15. TRAIN AND EVALUATE MODELS
# ============================================================

X_train_selected = X_train[selected_features]
X_test_selected = X_test[selected_features]

results = []
trained_models = {}
predictions_dict = {}

for name, estimator in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            estimator
        )
    ])

    pipeline.fit(
        X_train_selected,
        y_train
    )

    predictions = pipeline.predict(
        X_test_selected
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    trained_models[name] = pipeline
    predictions_dict[name] = predictions

    print("Accuracy:", round(accuracy, 4))
    print("F1 Score:", round(f1, 4))

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    output_path("model_comparison.csv"),
    index=False
)

print("\nMODEL COMPARISON:")
print(comparison_df.to_string(index=False))

# ============================================================
# 16. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    hue="Model",
    palette="viridis",
    legend=False
)

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

plt.title(
    "AIS Model Accuracy Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Accuracy")
plt.ylim(0, 1.12)
plt.xticks(rotation=30, ha="right")

save_show("accuracy_graph.png")

# ============================================================
# 17. CORRELATION HEATMAP
# ============================================================

correlation_columns = (
    selected_features +
    ["Total_Screening_Percentage"]
)

correlation = df[
    correlation_columns
].corr(numeric_only=True)

plt.figure(figsize=(13, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title(
    "AIS Selected Features Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

save_show("heatmap.png")

# ============================================================
# 18. COMPARISON GRAPH
# ============================================================

comparison_long = comparison_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=comparison_long,
    x="Model",
    y="Score",
    hue="Metric",
    palette="Set2"
)

plt.title(
    "AIS Model Performance Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel("Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=30, ha="right")
plt.legend(title="Metric")

save_show("comparison_graph.png")

# ============================================================
# 19. PRIMARY AIS MODEL
# ============================================================

# Always report the AIS Random Forest as
# the primary model. Do not select it based
# on held-out test performance.

best_model_name = "AIS Random Forest"

best_model = trained_models[
    best_model_name
]

best_predictions = predictions_dict[
    best_model_name
]

print("\nAIS Classification Report:")
print(
    classification_report(
        y_test,
        best_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 20. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    best_predictions,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "AIS Random Forest Confusion Matrix",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

save_show("confusion_matrix.png")

# ============================================================
# 21. RESULT CSV
# ============================================================

result_df = df.loc[
    X_test.index,
    [
        "State_UT",
        "Total_Screening_Percentage",
        "Total_Detection_Percentage"
    ]
].copy()

result_df["Actual_Category"] = (
    encoder.inverse_transform(y_test)
)

result_df["Predicted_Category"] = (
    encoder.inverse_transform(
        best_predictions
    )
)

result_df["Actual_Encoded"] = y_test
result_df["Predicted_Encoded"] = best_predictions

result_df["Correct_Prediction"] = (
    result_df["Actual_Category"]
    == result_df["Predicted_Category"]
)

result_df = result_df.sort_values(
    "State_UT"
).reset_index(drop=True)

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\nAIS RESULT CSV:")
print(result_df.to_string(index=False))

# ============================================================
# 22. RESULT GRAPH
# ============================================================

positions = np.arange(
    len(result_df)
)

plt.figure(figsize=(14, 7))

plt.plot(
    positions,
    result_df["Actual_Encoded"],
    marker="o",
    linewidth=2,
    label="Actual"
)

plt.plot(
    positions,
    result_df["Predicted_Encoded"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Predicted"
)

plt.title(
    "AIS Actual vs Predicted Screening Performance",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("State / UT")
plt.ylabel("Screening Category")

plt.xticks(
    positions,
    result_df["State_UT"],
    rotation=75,
    ha="right"
)

plt.yticks(
    range(num_classes),
    class_names
)

plt.legend()
plt.grid(alpha=0.3)

save_show("result_graph.png")

# ============================================================
# 23. TRAIN FINAL AIS MODEL ON ALL DATA
# ============================================================

# Refit using the optimized features and parameters.
# Full-data predictions are fitted predictions,
# not independent validation results.

final_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=best_antibody["n_estimators"],
            max_depth=best_antibody["max_depth"],
            min_samples_leaf=best_antibody["min_samples_leaf"],
            class_weight="balanced",
            random_state=SEED
        )
    )
])

final_model.fit(
    X[selected_features],
    y
)

all_predictions = final_model.predict(
    X[selected_features]
)

all_probabilities = final_model.predict_proba(
    X[selected_features]
)

# ============================================================
# 24. PREDICTION CSV
# ============================================================

prediction_df = df[
    [
        "State_UT",
        "Total_Screening_Percentage",
        "Total_Detection_Percentage"
    ]
].copy()

prediction_df["Predicted_Category"] = (
    encoder.inverse_transform(
        all_predictions
    )
)

prediction_df["Predicted_Encoded"] = (
    all_predictions
)

prediction_df["Prediction_Confidence"] = (
    np.max(all_probabilities, axis=1)
)

model_classes = (
    final_model.named_steps[
        "classifier"
    ].classes_
)

for i, class_id in enumerate(model_classes):

    class_name = encoder.inverse_transform(
        [class_id]
    )[0]

    prediction_df[
        f"Probability_{class_name}"
    ] = all_probabilities[:, i]

prediction_df = prediction_df.sort_values(
    "State_UT"
).reset_index(drop=True)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\nAIS PREDICTIONS:")
print(prediction_df.to_string(index=False))

# ============================================================
# 25. PREDICTION GRAPH
# ============================================================

category_colors = {
    "Low": "tomato",
    "Moderate": "orange",
    "High": "seagreen"
}

colors = [
    category_colors.get(
        category,
        "steelblue"
    )
    for category in prediction_df[
        "Predicted_Category"
    ]
]

plt.figure(figsize=(15, 8))

plt.bar(
    prediction_df["State_UT"],
    prediction_df["Predicted_Encoded"],
    color=colors
)

plt.title(
    "AIS State-Level Screening Performance Predictions",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Predicted Category")

plt.yticks(
    range(num_classes),
    class_names
)

plt.xticks(
    rotation=85,
    ha="right"
)

save_show("prediction_graph.png")

# ============================================================
# 26. FEATURE IMPORTANCE GRAPH
# ============================================================

rf_classifier = (
    final_model.named_steps["classifier"]
)

feature_importance_df = pd.DataFrame({
    "Feature": selected_features,
    "Importance": rf_classifier.feature_importances_
})

feature_importance_df = (
    feature_importance_df.sort_values(
        "Importance",
        ascending=False
    )
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=feature_importance_df,
    x="Importance",
    y="Feature",
    hue="Feature",
    palette="viridis",
    legend=False
)

plt.title(
    "AIS Optimized Feature Importance",
    fontsize=15,
    fontweight="bold"
)

save_show("feature_importance_graph.png")

feature_importance_df.to_csv(
    output_path("feature_importance.csv"),
    index=False
)

# ============================================================
# 27. CLASS DISTRIBUTION GRAPH
# ============================================================

class_distribution = (
    df["Screening_Performance_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

plt.figure(figsize=(9, 6))

ax = sns.barplot(
    x=class_distribution.index,
    y=class_distribution.values,
    hue=class_distribution.index,
    palette="Set2",
    legend=False
)

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%d",
        padding=3
    )

plt.title(
    "AIS Screening Performance Class Distribution",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Category")
plt.ylabel("Number of States / UTs")

save_show("class_distribution_graph.png")

# ============================================================
# 28. SAVE OPTIMIZED MODEL
# ============================================================

model_package = {
    "model": final_model,
    "label_encoder": encoder,
    "selected_features": selected_features,
    "all_features": feature_columns,
    "best_antibody": best_antibody,
    "class_names": class_names,
    "target_column":
        "Screening_Performance_Category"
}

joblib.dump(
    model_package,
    output_path("model.pkl")
)

print("\nAIS model saved successfully!")

# ============================================================
# 29. SAVE AIS YAML CONFIGURATION
# ============================================================

config = {
    "project":
        "AI-Based TB-COVID Co-Infection Risk Analysis",

    "optimization":
        "Artificial Immune System - Clonal Selection",

    "dataset": DATA_PATH,

    "random_state": SEED,

    "ais_parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "top_antibodies": TOP_ANTIBODIES,
        "clones_per_antibody": CLONES_PER_ANTIBODY,
        "random_new_antibodies": RANDOM_NEW_ANTIBODIES,
        "cv_folds": CV_FOLDS
    },

    "best_parameters": {
        "n_estimators":
            best_antibody["n_estimators"],

        "max_depth":
            best_antibody["max_depth"],

        "min_samples_leaf":
            best_antibody["min_samples_leaf"]
    },

    "selected_features": selected_features,

    "class_names": class_names,

    "best_fitness": float(best_fitness),

    "test_accuracy": float(
        accuracy_score(
            y_test,
            best_predictions
        )
    )
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ============================================================
# 30. SAVE AIS JSON METADATA
# ============================================================

metadata = {
    "project":
        "AI-Based TB-COVID Co-Infection Risk Analysis",

    "algorithm":
        "Artificial Immune System (CLONALG)",

    "created_at":
        datetime.now().isoformat(),

    "dataset_records": int(len(df)),

    "training_records": int(len(X_train)),

    "testing_records": int(len(X_test)),

    "selected_features": selected_features,

    "number_of_selected_features":
        len(selected_features),

    "best_fitness": float(best_fitness),

    "optimization_history": [
        float(value)
        for value in fitness_history
    ],

    "test_metrics": {
        "accuracy": float(
            accuracy_score(
                y_test,
                best_predictions
            )
        ),

        "precision": float(
            precision_score(
                y_test,
                best_predictions,
                average="weighted",
                zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y_test,
                best_predictions,
                average="weighted",
                zero_division=0
            )
        ),

        "f1_score": float(
            f1_score(
                y_test,
                best_predictions,
                average="weighted",
                zero_division=0
            )
        )
    },

    "class_names": class_names,

    "confusion_matrix": cm.tolist(),

    "limitations": [
        "Small state-level dataset",
        "Screening labels are derived from coverage",
        "Results are exploratory",
        "Not a clinical risk prediction model"
    ]
}

with open(
    output_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

# ============================================================
# 31. VERIFY GENERATED OUTPUTS
# ============================================================

output_files = [
    "model.pkl",
    "config.yaml",
    "metadata.json",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "feature_importance_graph.png",
    "feature_importance.csv",
    "class_distribution_graph.png",
    "model_comparison.csv",
    "optimization_graph.png",
    "selected_features.csv"
]

print("\n" + "=" * 75)
print("VERIFYING AIS OUTPUT FILES")
print("=" * 75)

for filename in output_files:

    full_path = output_path(filename)

    if os.path.exists(full_path):

        size_kb = (
            os.path.getsize(full_path) / 1024
        )

        print(
            f"[SUCCESS] ais_{filename} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[MISSING] ais_{filename}"
        )

# ============================================================
# 32. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("=" * 75)

print("\nSelected Features:")
print(selected_features)

print("\nBest AIS Parameters:")
print({
    "n_estimators":
        best_antibody["n_estimators"],

    "max_depth":
        best_antibody["max_depth"],

    "min_samples_leaf":
        best_antibody["min_samples_leaf"]
})

print("\nAIS Test Accuracy:")
print(
    round(
        accuracy_score(
            y_test,
            best_predictions
        ),
        4
    )
)

print("\nModel Comparison:")
print(comparison_df.to_string(index=False))

print("\nAll AIS outputs saved in:")
print(BASE_DIR)

print("\nDONE!")
```
