```python
import os
import json
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based TB-COVID Co-Infection Risk Analysis"
)

DATA_PATH = os.path.join(
    BASE_DIR,
    "2.13_TB_COVID.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

SEED = 42
rng = np.random.default_rng(SEED)

np.random.seed(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 120


def output_path(filename):
    return os.path.join(BASE_DIR, "pso_" + filename)


def save_show(filename):
    path = output_path(filename)

    plt.tight_layout()
    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print("Saved:", path)


print("=" * 75)
print("PSO-BASED TB-COVID CO-INFECTION RISK ANALYSIS")
print("=" * 75)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(
    DATA_PATH,
    encoding="utf-8-sig",
    index_col=False
)

df.columns = df.columns.str.strip()

print("\nDataset loaded successfully!")
print("Original shape:", df.shape)

print("\nFirst five records:")
print(df.head())

print("\nOriginal columns:")
print(df.columns.tolist())

# ============================================================
# 3. STANDARDIZE COLUMN NAMES
# ============================================================

# Assumes the same 13-column order used in
# the earlier TB-COVID scripts.
# Check the printed original headers to verify.

if df.shape[1] != 13:
    raise ValueError(
        "Expected 13 columns but found "
        f"{df.shape[1]}."
    )

df.columns = [
    "State_UT",
    "Public_Screened_Count",
    "Public_Screening_Percentage",
    "Private_Screened_Count",
    "Private_Screening_Percentage",
    "Total_Screened_Count",
    "Total_Screening_Percentage",
    "Public_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detected_Count",
    "Private_Detection_Percentage",
    "Total_Detected_Count",
    "Total_Detection_Percentage"
]

# ============================================================
# 4. DATA CLEANING
# ============================================================

df["State_UT"] = (
    df["State_UT"]
    .astype(str)
    .str.strip()
)

numeric_columns = [
    col for col in df.columns
    if col != "State_UT"
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

print("\nMissing values:")
print(df.isnull().sum())

df = df.dropna(
    subset=["Total_Screening_Percentage"]
).copy()

aggregate_names = {
    "india",
    "all india",
    "total",
    "grand total"
}

df = df[
    ~df["State_UT"]
    .str.lower()
    .isin(aggregate_names)
].copy()

df.reset_index(drop=True, inplace=True)

print("\nCleaned dataset shape:", df.shape)

# ============================================================
# 5. CREATE CLASSIFICATION TARGET
# ============================================================

# Illustrative screening coverage thresholds.
# These are not clinical risk thresholds.

def classify_screening(value):

    if value < 50:
        return "Low"

    elif value < 80:
        return "Moderate"

    return "High"


df["Screening_Performance_Category"] = (
    df["Total_Screening_Percentage"]
    .apply(classify_screening)
)

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df["Screening_Performance_Category"]
)

class_names = label_encoder.classes_.tolist()
num_classes = len(class_names)

print("\nScreening categories:")
print(class_names)

print("\nClass distribution:")
print(
    df["Screening_Performance_Category"]
    .value_counts()
)

if num_classes < 2:
    raise ValueError(
        "At least two categories are required."
    )

# ============================================================
# 6. SELECT INPUT FEATURES
# ============================================================

# Target-defining and related screening percentages
# are excluded from the predictive features.
#
# This project models state-level screening
# performance, not patient infection risk.

feature_columns = [
    "Public_Screened_Count",
    "Private_Screened_Count",
    "Total_Screened_Count",
    "Public_Detected_Count",
    "Private_Detected_Count",
    "Total_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detection_Percentage",
    "Total_Detection_Percentage"
]

X = df[feature_columns].copy()

n_features = len(feature_columns)

print("\nAvailable features:")
for feature in feature_columns:
    print("-", feature)

# ============================================================
# 7. TRAIN-TEST SPLIT
# ============================================================

class_counts = np.bincount(y)

test_count = max(
    num_classes,
    int(np.ceil(len(y) * 0.25))
)

if test_count >= len(y):
    raise ValueError(
        "Insufficient data for train-test split."
    )

stratify_target = None

if (
    class_counts.min() >= 2
    and len(y) - test_count >= num_classes
):
    stratify_target = y

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=test_count,
        random_state=SEED,
        stratify=stratify_target
    )
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 8. PSO CONFIGURATION
# ============================================================

N_PARTICLES = 12
N_ITERATIONS = 15

INERTIA_WEIGHT = 0.7
COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

# Particle representation:
#
# Positions 0 to n_features-1:
# Continuous feature-selection values [0, 1]
# Feature selected when position >= 0.5
#
# Last 3 dimensions:
# n_estimators, max_depth, min_samples_leaf

DIMENSIONS = n_features + 3

LOWER_BOUNDS = np.concatenate([
    np.zeros(n_features),
    np.array([40, 2, 1])
])

UPPER_BOUNDS = np.concatenate([
    np.ones(n_features),
    np.array([150, 10, 4])
])

VELOCITY_LIMITS = (
    UPPER_BOUNDS - LOWER_BOUNDS
) * 0.2

# ============================================================
# 9. DECODE PARTICLE
# ============================================================

def decode_particle(position):

    mask = (
        position[:n_features] >= 0.5
    ).astype(int)

    # Ensure at least one feature
    if mask.sum() == 0:

        best_index = np.argmax(
            position[:n_features]
        )

        mask[best_index] = 1

    n_estimators = int(
        np.clip(
            np.rint(position[n_features]),
            40,
            150
        )
    )

    max_depth = int(
        np.clip(
            np.rint(position[n_features + 1]),
            2,
            10
        )
    )

    min_samples_leaf = int(
        np.clip(
            np.rint(position[n_features + 2]),
            1,
            4
        )
    )

    return {
        "mask": mask,
        "n_estimators": n_estimators,
        "max_depth": max_depth,
        "min_samples_leaf": min_samples_leaf
    }

# ============================================================
# 10. FITNESS FUNCTION
# ============================================================

# Optimize macro F1-score using only training data.
# The test set is reserved for final evaluation.

train_class_counts = np.bincount(
    y_train,
    minlength=num_classes
)

positive_counts = train_class_counts[
    train_class_counts > 0
]

CV_FOLDS = min(
    3,
    int(positive_counts.min())
)

if CV_FOLDS < 2:
    raise ValueError(
        "Not enough training samples per class "
        "for stratified PSO optimization. "
        "More observations are required."
    )

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}


def calculate_fitness(position):

    particle = decode_particle(position)

    mask = particle["mask"]

    selected_indices = np.where(
        mask == 1
    )[0]

    selected_features = [
        feature_columns[i]
        for i in selected_indices
    ]

    key = (
        tuple(mask.tolist()),
        particle["n_estimators"],
        particle["max_depth"],
        particle["min_samples_leaf"]
    )

    if key in fitness_cache:
        return fitness_cache[key]

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=particle["n_estimators"],
                max_depth=particle["max_depth"],
                min_samples_leaf=particle["min_samples_leaf"],
                class_weight="balanced",
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

    try:

        scores = cross_val_score(
            pipeline,
            X_train[selected_features],
            y_train,
            cv=cv,
            scoring="f1_macro",
            n_jobs=1,
            error_score="raise"
        )

        mean_f1 = float(np.mean(scores))

    except ValueError:

        mean_f1 = 0.0

    # Encourage compact feature subsets
    feature_penalty = (
        0.02 *
        len(selected_features) /
        n_features
    )

    fitness = mean_f1 - feature_penalty

    fitness_cache[key] = fitness

    return fitness

# ============================================================
# 11. INITIALIZE PARTICLE SWARM
# ============================================================

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(N_PARTICLES, DIMENSIONS)
)

velocities = rng.uniform(
    -VELOCITY_LIMITS,
    VELOCITY_LIMITS,
    size=(N_PARTICLES, DIMENSIONS)
)

personal_best_positions = positions.copy()

personal_best_scores = np.full(
    N_PARTICLES,
    -np.inf
)

global_best_position = None
global_best_score = -np.inf

optimization_history = []

print("\nInitializing PSO particles...")

# ============================================================
# 12. PARTICLE SWARM OPTIMIZATION
# ============================================================

print("\nStarting PSO optimization...")

for iteration in range(N_ITERATIONS):

    # Evaluate each particle
    for i in range(N_PARTICLES):

        score = calculate_fitness(
            positions[i]
        )

        if score > personal_best_scores[i]:

            personal_best_scores[i] = score

            personal_best_positions[i] = (
                positions[i].copy()
            )

        if score > global_best_score:

            global_best_score = score

            global_best_position = (
                positions[i].copy()
            )

    optimization_history.append(
        global_best_score
    )

    print(
        f"Iteration {iteration + 1}/{N_ITERATIONS}"
        f" | Best Fitness: {global_best_score:.4f}"
    )

    # Update velocity and position
    for i in range(N_PARTICLES):

        r1 = rng.random(DIMENSIONS)
        r2 = rng.random(DIMENSIONS)

        cognitive_velocity = (
            COGNITIVE_COEFFICIENT
            * r1
            * (
                personal_best_positions[i]
                - positions[i]
            )
        )

        social_velocity = (
            SOCIAL_COEFFICIENT
            * r2
            * (
                global_best_position
                - positions[i]
            )
        )

        velocities[i] = (
            INERTIA_WEIGHT * velocities[i]
            + cognitive_velocity
            + social_velocity
        )

        velocities[i] = np.clip(
            velocities[i],
            -VELOCITY_LIMITS,
            VELOCITY_LIMITS
        )

        positions[i] += velocities[i]

        positions[i] = np.clip(
            positions[i],
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )

print("\nPSO optimization completed!")

print("Best fitness:", global_best_score)

# ============================================================
# 13. DECODE BEST PARTICLE
# ============================================================

best_particle = decode_particle(
    global_best_position
)

selected_indices = np.where(
    best_particle["mask"] == 1
)[0]

selected_features = [
    feature_columns[i]
    for i in selected_indices
]

best_params = {
    "n_estimators":
        best_particle["n_estimators"],

    "max_depth":
        best_particle["max_depth"],

    "min_samples_leaf":
        best_particle["min_samples_leaf"]
}

print("\nBest PSO parameters:")
print(best_params)

print("\nPSO selected features:")
for feature in selected_features:
    print("-", feature)

# ============================================================
# 14. SAVE SELECTED FEATURES CSV
# ============================================================

selected_features_df = pd.DataFrame({
    "Feature": feature_columns,
    "Selected": best_particle["mask"].astype(int)
})

selected_features_df.to_csv(
    output_path("selected_features.csv"),
    index=False
)

# ============================================================
# 15. SAVE OPTIMIZATION HISTORY
# ============================================================

optimization_df = pd.DataFrame({
    "Iteration": np.arange(
        1,
        N_ITERATIONS + 1
    ),
    "Best_Fitness": optimization_history
})

optimization_df.to_csv(
    output_path("optimization_history.csv"),
    index=False
)

# ============================================================
# 16. PSO OPTIMIZATION GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    optimization_df["Iteration"],
    optimization_df["Best_Fitness"],
    marker="o",
    linewidth=2,
    color="purple"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Iteration")
plt.ylabel("Best Fitness (Macro F1 - Feature Penalty)")
plt.grid(alpha=0.3)

save_show("optimization_graph.png")

# ============================================================
# 17. DEFINE MACHINE LEARNING MODELS
# ============================================================

models = {
    "PSO Random Forest": RandomForestClassifier(
        n_estimators=best_params["n_estimators"],
        max_depth=best_params["max_depth"],
        min_samples_leaf=best_params["min_samples_leaf"],
        class_weight="balanced",
        random_state=SEED
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=4,
        class_weight="balanced",
        random_state=SEED
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=SEED
    ),

    "SVM": SVC(
        kernel="rbf",
        probability=True,
        class_weight="balanced",
        random_state=SEED
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=3
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        random_state=SEED
    )
}

# ============================================================
# 18. TRAIN AND EVALUATE MODELS
# ============================================================

X_train_selected = X_train[
    selected_features
]

X_test_selected = X_test[
    selected_features
]

results = []
trained_models = {}
predictions_dict = {}

for name, estimator in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            estimator
        )
    ])

    pipeline.fit(
        X_train_selected,
        y_train
    )

    predictions = pipeline.predict(
        X_test_selected
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    trained_models[name] = pipeline
    predictions_dict[name] = predictions

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

# ============================================================
# 19. MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by=["F1_Score", "Accuracy"],
    ascending=False
).reset_index(drop=True)

comparison_df.to_csv(
    output_path("model_comparison.csv"),
    index=False
)

print("\nMODEL COMPARISON")
print(comparison_df.to_string(index=False))

# ============================================================
# 20. ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Model",
    y="Accuracy",
    hue="Model",
    palette="viridis",
    legend=False
)

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%.3f",
        padding=3
    )

plt.title(
    "PSO-Based Machine Learning Accuracy Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")
plt.ylim(0, 1.12)
plt.xticks(rotation=30, ha="right")

save_show("accuracy_graph.png")

# ============================================================
# 21. CORRELATION HEATMAP
# ============================================================

heatmap_columns = (
    selected_features
    + ["Total_Screening_Percentage"]
)

correlation_matrix = df[
    heatmap_columns
].corr(numeric_only=True)

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title(
    "PSO Selected Features Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

save_show("heatmap.png")

# ============================================================
# 22. PERFORMANCE COMPARISON GRAPH
# ============================================================

comparison_long = comparison_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=comparison_long,
    x="Model",
    y="Score",
    hue="Metric",
    palette="Set2"
)

plt.title(
    "PSO Model Performance Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Performance Score")
plt.ylim(0, 1.1)
plt.xticks(rotation=30, ha="right")
plt.legend(title="Metric")

save_show("comparison_graph.png")

# ============================================================
# 23. PRIMARY PSO MODEL
# ============================================================

# Use PSO Random Forest as the primary model.
# Do not select a model based on the test scores.

primary_model_name = "PSO Random Forest"

primary_model = trained_models[
    primary_model_name
]

primary_predictions = predictions_dict[
    primary_model_name
]

print("\nPSO Classification Report:")

print(
    classification_report(
        y_test,
        primary_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 24. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    primary_predictions,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "PSO Random Forest Confusion Matrix",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")

save_show("confusion_matrix.png")

# ============================================================
# 25. CREATE RESULT CSV
# ============================================================

# Held-out test-set predictions only.

result_df = df.loc[
    X_test.index,
    [
        "State_UT",
        "Total_Screening_Percentage",
        "Total_Detection_Percentage"
    ]
].copy()

result_df["Actual_Category"] = (
    label_encoder.inverse_transform(y_test)
)

result_df["Predicted_Category"] = (
    label_encoder.inverse_transform(
        primary_predictions
    )
)

result_df["Actual_Encoded"] = y_test

result_df["Predicted_Encoded"] = (
    primary_predictions
)

result_df["Correct_Prediction"] = (
    result_df["Actual_Category"]
    == result_df["Predicted_Category"]
)

result_df = result_df.sort_values(
    "State_UT"
).reset_index(drop=True)

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\nPSO RESULT CSV")
print(result_df.to_string(index=False))

# ============================================================
# 26. RESULT GRAPH
# ============================================================

positions = np.arange(
    len(result_df)
)

plt.figure(figsize=(14, 7))

plt.plot(
    positions,
    result_df["Actual_Encoded"],
    marker="o",
    linewidth=2,
    label="Actual"
)

plt.plot(
    positions,
    result_df["Predicted_Encoded"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted"
)

plt.title(
    "PSO Actual vs Predicted Screening Performance",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Screening Performance Category")

plt.xticks(
    positions,
    result_df["State_UT"],
    rotation=75,
    ha="right"
)

plt.yticks(
    range(num_classes),
    class_names
)

plt.legend()
plt.grid(alpha=0.3)

save_show("result_graph.png")

# ============================================================
# 27. TRAIN FINAL MODEL ON ALL DATA
# ============================================================

# This model is used for full-dataset fitted
# predictions after test evaluation.
#
# It uses the PSO-selected feature subset and
# optimized hyperparameters.

final_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=best_params["n_estimators"],
            max_depth=best_params["max_depth"],
            min_samples_leaf=best_params["min_samples_leaf"],
            class_weight="balanced",
            random_state=SEED
        )
    )
])

final_model.fit(
    X[selected_features],
    y
)

all_predictions = final_model.predict(
    X[selected_features]
)

all_probabilities = final_model.predict_proba(
    X[selected_features]
)

# ============================================================
# 28. CREATE PREDICTION CSV
# ============================================================

prediction_df = df[
    [
        "State_UT",
        "Total_Screening_Percentage",
        "Total_Detection_Percentage"
    ]
].copy()

prediction_df["Predicted_Category"] = (
    label_encoder.inverse_transform(
        all_predictions
    )
)

prediction_df["Predicted_Encoded"] = (
    all_predictions
)

prediction_df["Prediction_Confidence"] = (
    np.max(all_probabilities, axis=1)
)

model_classes = (
    final_model.named_steps["classifier"].classes_
)

for i, class_id in enumerate(model_classes):

    class_name = label_encoder.inverse_transform(
        [class_id]
    )[0]

    prediction_df[
        f"Probability_{class_name}"
    ] = all_probabilities[:, i]

prediction_df = prediction_df.sort_values(
    "State_UT"
).reset_index(drop=True)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\nPSO PREDICTION CSV")
print(prediction_df.to_string(index=False))

# ============================================================
# 29. PREDICTION GRAPH
# ============================================================

category_colors = {
    "Low": "tomato",
    "Moderate": "orange",
    "High": "seagreen"
}

colors = [
    category_colors.get(
        category,
        "steelblue"
    )
    for category in prediction_df[
        "Predicted_Category"
    ]
]

plt.figure(figsize=(16, 8))

plt.bar(
    prediction_df["State_UT"],
    prediction_df["Predicted_Encoded"],
    color=colors
)

plt.title(
    "PSO State-Level Screening Performance Predictions",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Predicted Category")

plt.yticks(
    range(num_classes),
    class_names
)

plt.xticks(
    rotation=85,
    ha="right"
)

save_show("prediction_graph.png")

# ============================================================
# 30. FEATURE IMPORTANCE
# ============================================================

rf_classifier = (
    final_model.named_steps["classifier"]
)

feature_importance_df = pd.DataFrame({
    "Feature": selected_features,
    "Importance": rf_classifier.feature_importances_
})

feature_importance_df = (
    feature_importance_df.sort_values(
        "Importance",
        ascending=False
    )
)

feature_importance_df.to_csv(
    output_path("feature_importance.csv"),
    index=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=feature_importance_df,
    x="Importance",
    y="Feature",
    hue="Feature",
    palette="viridis",
    legend=False
)

plt.title(
    "PSO Optimized Feature Importance",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Feature Importance")
plt.ylabel("Selected Features")

save_show("feature_importance_graph.png")

# ============================================================
# 31. CLASS DISTRIBUTION GRAPH
# ============================================================

class_distribution = (
    df["Screening_Performance_Category"]
    .value_counts()
    .reindex(class_names, fill_value=0)
)

plt.figure(figsize=(9, 6))

ax = sns.barplot(
    x=class_distribution.index,
    y=class_distribution.values,
    hue=class_distribution.index,
    palette="Set2",
    legend=False
)

for container in ax.containers:

    ax.bar_label(
        container,
        fmt="%d",
        padding=3
    )

plt.title(
    "PSO Screening Performance Class Distribution",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Screening Performance Category")
plt.ylabel("Number of States / UTs")

save_show("class_distribution_graph.png")

# ============================================================
# 32. SAVE TRAINED MODEL
# ============================================================

model_package = {
    "model": final_model,
    "label_encoder": label_encoder,
    "selected_features": selected_features,
    "all_features": feature_columns,
    "best_parameters": best_params,
    "class_names": class_names,
    "target_column":
        "Screening_Performance_Category"
}

joblib.dump(
    model_package,
    output_path("model.pkl")
)

print("\nPSO model saved successfully!")

# ============================================================
# 33. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project_name":
        "AI-Based TB-COVID Co-Infection Risk Analysis",

    "optimization_method":
        "Particle Swarm Optimization",

    "dataset": DATA_PATH,

    "random_state": SEED,

    "pso_parameters": {
        "particles": N_PARTICLES,
        "iterations": N_ITERATIONS,
        "inertia_weight": INERTIA_WEIGHT,
        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,
        "social_coefficient":
            SOCIAL_COEFFICIENT,
        "cv_folds": CV_FOLDS
    },

    "best_parameters": best_params,

    "selected_features": selected_features,

    "class_names": class_names,

    "best_fitness": float(
        global_best_score
    ),

    "test_accuracy": float(
        accuracy_score(
            y_test,
            primary_predictions
        )
    )
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("\nPSO configuration saved!")

# ============================================================
# 34. SAVE JSON METADATA
# ============================================================

metadata = {
    "project_name":
        "AI-Based TB-COVID Co-Infection Risk Analysis",

    "optimization":
        "Particle Swarm Optimization",

    "created_at":
        datetime.now().isoformat(),

    "dataset_records": int(len(df)),

    "training_records": int(len(X_train)),

    "testing_records": int(len(X_test)),

    "selected_features": selected_features,

    "number_of_selected_features":
        len(selected_features),

    "best_parameters": best_params,

    "best_fitness": float(
        global_best_score
    ),

    "optimization_history": [
        float(value)
        for value in optimization_history
    ],

    "test_metrics": {
        "accuracy": float(
            accuracy_score(
                y_test,
                primary_predictions
            )
        ),

        "precision": float(
            precision_score(
                y_test,
                primary_predictions,
                average="weighted",
                zero_division=0
            )
        ),

        "recall": float(
            recall_score(
                y_test,
                primary_predictions,
                average="weighted",
                zero_division=0
            )
        ),

        "f1_score": float(
            f1_score(
                y_test,
                primary_predictions,
                average="weighted",
                zero_division=0
            )
        )
    },

    "class_names": class_names,

    "confusion_matrix": cm.tolist(),

    "limitations": [
        "Small state-level dataset",
        "Screening classes derived from coverage",
        "Exploratory model evaluation only",
        "Not a clinical risk prediction system"
    ]
}

with open(
    output_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\nPSO metadata saved!")

# ============================================================
# 35. VERIFY GENERATED FILES
# ============================================================

output_files = [
    "model.pkl",
    "config.yaml",
    "metadata.json",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "optimization_graph.png",
    "feature_importance_graph.png",
    "feature_importance.csv",
    "class_distribution_graph.png",
    "model_comparison.csv",
    "selected_features.csv",
    "optimization_history.csv"
]

print("\n" + "=" * 75)
print("VERIFYING PSO OUTPUT FILES")
print("=" * 75)

for filename in output_files:

    full_path = output_path(filename)

    if os.path.exists(full_path):

        size_kb = (
            os.path.getsize(full_path) / 1024
        )

        print(
            f"[SUCCESS] pso_{filename} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[MISSING] pso_{filename}"
        )

# ============================================================
# 36. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("PSO PROJECT EXECUTION COMPLETED")
print("=" * 75)

print("\nBest PSO Fitness:")
print(round(global_best_score, 4))

print("\nSelected Features:")
print(selected_features)

print("\nOptimized Parameters:")
print(best_params)

print("\nPSO Test Accuracy:")
print(
    round(
        accuracy_score(
            y_test,
            primary_predictions
        ),
        4
    )
)

print("\nModel Comparison:")
print(comparison_df.to_string(index=False))

print("\nAll PSO results saved in:")
print(BASE_DIR)

print("\nDONE!")
```
