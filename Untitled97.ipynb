```python
import os
import json
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import joblib
import yaml
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based TB-COVID Co-Infection Risk Analysis"
)

DATA_PATH = os.path.join(
    BASE_DIR,
    "2.13_TB_COVID.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

H5_PATH = os.path.join(
    BASE_DIR, "tb_covid_model.h5"
)

PKL_PATH = os.path.join(
    BASE_DIR, "tb_covid_model.pkl"
)

YAML_PATH = os.path.join(
    BASE_DIR, "tb_covid_config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR, "tb_covid_metadata.json"
)

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

print("=" * 65)
print("AI-BASED TB-COVID CO-INFECTION RISK ANALYSIS")
print("=" * 65)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(
    DATA_PATH,
    encoding="utf-8-sig",
    index_col=False
)

# Clean column names
df.columns = df.columns.str.strip()

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
print(df.head())

print("\nDataset information:")
df.info()

# ============================================================
# 3. RENAME DATASET COLUMNS
# ============================================================

# The supplied CSV has 13 columns in this exact order.

if df.shape[1] != 13:
    raise ValueError(
        "Expected 13 columns. "
        f"Found {df.shape[1]} columns."
    )

df.columns = [
    "State_UT",
    "Public_Screened_Count",
    "Public_Screening_Percentage",
    "Private_Screened_Count",
    "Private_Screening_Percentage",
    "Total_Screened_Count",
    "Total_Screening_Percentage",
    "Public_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detected_Count",
    "Private_Detection_Percentage",
    "Total_Detected_Count",
    "Total_Detection_Percentage"
]

print("\nRenamed columns:")
print(df.columns.tolist())

# ============================================================
# 4. DATA CLEANING
# ============================================================

df["State_UT"] = (
    df["State_UT"]
    .astype(str)
    .str.strip()
)

numeric_columns = [
    col for col in df.columns
    if col != "State_UT"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip(),
        errors="coerce"
    )

print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Remove rows without target information
df = df.dropna(
    subset=["Total_Screening_Percentage"]
).copy()

# Do not train on aggregate totals, if present
aggregate_names = {
    "total",
    "grand total",
    "all india",
    "india"
}

df = df[
    ~df["State_UT"]
    .str.lower()
    .isin(aggregate_names)
].copy()

df = df.reset_index(drop=True)

print("\nRecords after cleaning:", len(df))

# ============================================================
# 5. CREATE SCREENING PERFORMANCE CATEGORIES
# ============================================================

# These are illustrative, predefined thresholds.
#
# Low:      screening coverage < 50%
# Moderate: screening coverage 50% to < 80%
# High:     screening coverage >= 80%
#
# They are not clinical risk thresholds.

def assign_screening_category(value):

    if value < 50:
        return "Low"

    elif value < 80:
        return "Moderate"

    else:
        return "High"


df["Screening_Performance_Category"] = (
    df["Total_Screening_Percentage"]
    .apply(assign_screening_category)
)

print("\nScreening performance distribution:")
print(
    df["Screening_Performance_Category"]
    .value_counts()
)

# ============================================================
# 6. FEATURE SELECTION
# ============================================================

# Do not include Total_Screening_Percentage:
# It directly defines the target category.
#
# Do not include State_UT:
# It is an identifier, not a numeric predictor.
#
# Other screening percentages are also excluded
# to reduce direct target leakage.
#
# Features describe screening volumes and
# observed TB-COVID co-infection detection.

feature_columns = [
    "Public_Screened_Count",
    "Private_Screened_Count",
    "Total_Screened_Count",
    "Public_Detected_Count",
    "Private_Detected_Count",
    "Total_Detected_Count",
    "Public_Detection_Percentage",
    "Private_Detection_Percentage",
    "Total_Detection_Percentage"
]

X = df[feature_columns].copy()

y_text = df[
    "Screening_Performance_Category"
].copy()

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(y_text)

class_names = label_encoder.classes_.tolist()
num_classes = len(class_names)

print("\nSelected features:")
print(feature_columns)

print("\nClass mapping:")
for i, name in enumerate(class_names):
    print(i, ":", name)

if num_classes < 2:
    raise ValueError(
        "Only one screening category exists. "
        "Classification requires at least two classes."
    )

# ============================================================
# 7. TRAIN-TEST SPLIT
# ============================================================

# The dataset is very small.
# Use stratification only when each class has
# enough observations.

class_counts = np.bincount(y)

test_count = max(
    num_classes,
    int(np.ceil(len(y) * 0.25))
)

test_count = min(
    test_count,
    len(y) - num_classes
)

if test_count <= 0:
    raise ValueError(
        "Not enough records for train-test split."
    )

stratify_target = None

if (
    class_counts.min() >= 2
    and test_count >= num_classes
    and len(y) - test_count >= num_classes
):
    stratify_target = y

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=test_count,
        random_state=RANDOM_STATE,
        stratify=stratify_target
    )
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# ============================================================
# 8. RANDOM FOREST MODEL
# ============================================================

print("\nTraining Random Forest...")

rf_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        RandomForestClassifier(
            n_estimators=100,
            max_depth=5,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=RANDOM_STATE
        )
    )
])

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)

print("\nRandom Forest Accuracy:", rf_accuracy)

print("\nRandom Forest Classification Report:")
print(
    classification_report(
        y_test,
        rf_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 9. SAVE RANDOM FOREST AS PKL
# ============================================================

# Save pipeline, encoder, and feature order
# together for future predictions.

model_package = {
    "model": rf_pipeline,
    "label_encoder": label_encoder,
    "feature_columns": feature_columns,
    "target_column":
        "Screening_Performance_Category",
    "class_names": class_names
}

joblib.dump(
    model_package,
    PKL_PATH
)

print("\nPKL model saved:")
print(PKL_PATH)

# ============================================================
# 10. PREPROCESS DATA FOR NEURAL NETWORK
# ============================================================

nn_imputer = SimpleImputer(
    strategy="median"
)

X_train_imputed = nn_imputer.fit_transform(
    X_train
)

X_test_imputed = nn_imputer.transform(
    X_test
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_imputed
)

X_test_scaled = scaler.transform(
    X_test_imputed
)

# Save neural network preprocessing separately.
# The H5 file stores the network architecture
# and weights, not the sklearn scaler.

nn_preprocessor_path = os.path.join(
    BASE_DIR,
    "tb_covid_nn_preprocessor.pkl"
)

joblib.dump(
    {
        "imputer": nn_imputer,
        "scaler": scaler,
        "label_encoder": label_encoder,
        "feature_columns": feature_columns
    },
    nn_preprocessor_path
)

# ============================================================
# 11. BUILD ARTIFICIAL NEURAL NETWORK
# ============================================================

print("\nBuilding Artificial Neural Network...")

nn_model = Sequential([
    Input(shape=(X_train_scaled.shape[1],)),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.2),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        num_classes,
        activation="softmax"
    )
])

nn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

nn_model.summary()

# ============================================================
# 12. TRAIN NEURAL NETWORK
# ============================================================

early_stopping = EarlyStopping(
    monitor="loss",
    patience=15,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_scaled,
    y_train,
    epochs=100,
    batch_size=8,
    callbacks=[early_stopping],
    verbose=1
)

# ============================================================
# 13. EVALUATE NEURAL NETWORK
# ============================================================

nn_probabilities = nn_model.predict(
    X_test_scaled,
    verbose=0
)

nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)

nn_accuracy = accuracy_score(
    y_test,
    nn_predictions
)

print("\nNeural Network Accuracy:", nn_accuracy)

print("\nNeural Network Classification Report:")
print(
    classification_report(
        y_test,
        nn_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 14. SAVE NEURAL NETWORK AS H5
# ============================================================

nn_model.save(H5_PATH)

print("\nH5 model saved:")
print(H5_PATH)

# ============================================================
# 15. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project_name":
        "AI-Based TB-COVID Co-Infection Risk Analysis",

    "dataset": {
        "file_name": "2.13_TB_COVID.csv",
        "path": DATA_PATH,
        "records_used": int(len(df)),
        "original_columns": 13
    },

    "target": {
        "column":
            "Screening_Performance_Category",

        "type": "multiclass_classification",

        "thresholds": {
            "low": "coverage < 50%",
            "moderate": "50% <= coverage < 80%",
            "high": "coverage >= 80%"
        },

        "classes": class_names
    },

    "features": feature_columns,

    "random_forest": {
        "n_estimators": 100,
        "max_depth": 5,
        "min_samples_leaf": 2,
        "random_state": RANDOM_STATE,
        "accuracy": float(rf_accuracy)
    },

    "neural_network": {
        "hidden_layers": [32, 16],
        "activation": "relu",
        "output_activation": "softmax",
        "optimizer": "adam",
        "loss": "sparse_categorical_crossentropy",
        "epochs_max": 100,
        "batch_size": 8,
        "accuracy": float(nn_accuracy)
    },

    "output_files": {
        "h5": H5_PATH,
        "pkl": PKL_PATH,
        "yaml": YAML_PATH,
        "json": JSON_PATH,
        "nn_preprocessor": nn_preprocessor_path
    }
}

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML configuration saved:")
print(YAML_PATH)

# ============================================================
# 16. SAVE JSON METADATA
# ============================================================

rf_cm = confusion_matrix(
    y_test,
    rf_predictions,
    labels=np.arange(num_classes)
)

nn_cm = confusion_matrix(
    y_test,
    nn_predictions,
    labels=np.arange(num_classes)
)

metadata = {
    "project_name":
        "AI-Based TB-COVID Co-Infection Risk Analysis",

    "created_at":
        datetime.now().isoformat(),

    "dataset": {
        "file": "2.13_TB_COVID.csv",
        "total_records_used": int(len(df)),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "number_of_features": len(feature_columns)
    },

    "features": feature_columns,

    "target":
        "Screening_Performance_Category",

    "class_names": class_names,

    "class_distribution": {
        str(key): int(value)
        for key, value in (
            df["Screening_Performance_Category"]
            .value_counts()
            .items()
        )
    },

    "models": {
        "Random_Forest": {
            "accuracy": float(rf_accuracy),
            "confusion_matrix": rf_cm.tolist()
        },

        "Neural_Network": {
            "accuracy": float(nn_accuracy),
            "confusion_matrix": nn_cm.tolist()
        }
    },

    "best_model_by_test_accuracy": (
        "Random_Forest"
        if rf_accuracy >= nn_accuracy
        else "Neural_Network"
    ),

    "limitations": [
        "Small state-level dataset",
        "Screening categories are derived from coverage",
        "No individual patient risk predictions",
        "Test accuracy is exploratory"
    ]
}

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\nJSON metadata saved:")
print(JSON_PATH)

# ============================================================
# 17. MODEL COMPARISON
# ============================================================

comparison_df = pd.DataFrame({
    "Model": [
        "Random Forest",
        "Artificial Neural Network"
    ],

    "Accuracy": [
        rf_accuracy,
        nn_accuracy
    ]
})

print("\nMODEL COMPARISON:")
print(comparison_df.to_string(index=False))

# ============================================================
# 18. VERIFY SAVED FILES
# ============================================================

print("\n" + "=" * 65)
print("VERIFYING SAVED FILES")
print("=" * 65)

saved_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,
    nn_preprocessor_path
]

for file_path in saved_files:

    if os.path.exists(file_path):

        size_kb = (
            os.path.getsize(file_path) / 1024
        )

        print(
            f"SUCCESS: {os.path.basename(file_path)} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            "MISSING:",
            file_path
        )

# ============================================================
# 19. TEST LOADING THE SAVED RANDOM FOREST
# ============================================================

print("\nTesting saved PKL model...")

loaded_package = joblib.load(PKL_PATH)

loaded_model = loaded_package["model"]

sample_data = X_test.iloc[[0]]

sample_prediction = loaded_model.predict(
    sample_data
)

predicted_category = (
    loaded_package["label_encoder"]
    .inverse_transform(sample_prediction)
)

actual_category = (
    label_encoder.inverse_transform(
        y_test[:1]
    )
)

print("\nSample State-Level Prediction:")
print("Predicted:", predicted_category[0])
print("Actual:", actual_category[0])

# ============================================================
# 20. TEST LOADING THE SAVED H5 MODEL
# ============================================================

print("\nTesting saved H5 model...")

loaded_nn = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)

loaded_preprocessor = joblib.load(
    nn_preprocessor_path
)

sample_nn_input = (
    loaded_preprocessor["imputer"]
    .transform(sample_data)
)

sample_nn_input = (
    loaded_preprocessor["scaler"]
    .transform(sample_nn_input)
)

sample_nn_probabilities = loaded_nn.predict(
    sample_nn_input,
    verbose=0
)

sample_nn_class = int(
    np.argmax(sample_nn_probabilities, axis=1)[0]
)

print(
    "Neural Network Prediction:",
    class_names[sample_nn_class]
)

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 65)

print("\nAll files saved in:")
print(BASE_DIR)
```
